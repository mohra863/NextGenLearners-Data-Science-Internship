## Step 2 — Explore the Database Tables

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("nextgen.db")

pd.read_sql("SELECT * FROM applicants LIMIT 5;", conn)
```

```python
# Explore the interns table
pd.read_sql("SELECT * FROM interns LIMIT 5;", conn)
```

```python
# Explore the hackathon_scores table
pd.read_sql("SELECT * FROM hackathon_scores LIMIT 5;", conn)
```

```python
# Verify the relationship between interns and hackathon_scores
query = """
SELECT
    i.intern_id,
    i.domain,
    i.completion_status,
    h.score
FROM interns i
JOIN hackathon_scores h
    ON i.intern_id = h.intern_id
ORDER BY h.score DESC;
"""

pd.read_sql(query, conn)
```

## Step 3 — Part A: SQL Queries

```python
# Query 1: Counts completed interns grouped by domain
# This query counts the number of interns who completed the program in each domain.
query1 = """
SELECT
    domain,
    COUNT(*) AS completed_count
FROM interns
WHERE completion_status = 'Completed'
GROUP BY domain
ORDER BY completed_count DESC;
"""

result1 = pd.read_sql(query1, conn)

result1
```

```python
# Query 2: Calculates the average hackathon score for each domain
# This query calculates the average hackathon score for each domain.
query2 = """
SELECT
    domain,
    ROUND(AVG(score), 2) AS avg_score
FROM hackathon_scores
GROUP BY domain
ORDER BY avg_score DESC;
"""

result2 = pd.read_sql(query2, conn)

result2
```

```python
# Query 3: Lists interns who scored 85 or above
# This query lists interns who scored 85 or above in the hackathon, as candidates for showcase or certificates.
query3 = """
SELECT
    i.intern_id,
    i.domain,
    h.score
FROM interns i
JOIN hackathon_scores h
    ON i.intern_id = h.intern_id
WHERE h.score >= 85
ORDER BY h.score DESC;
"""

result3 = pd.read_sql(query3, conn)

result3
```

```python
# Query 4: Calculates the conversion rate from applicants to completed interns
# This query compares the total number of applicants with the number of interns who completed the program in each domain.
query4 = """
SELECT
    a.domain,
    COUNT(DISTINCT a.applicant_id) AS total_applicants,
    COUNT(DISTINCT i.intern_id) AS total_completed,
    ROUND(
        100.0 * COUNT(DISTINCT i.intern_id)
        / COUNT(DISTINCT a.applicant_id),
        2
    ) AS conversion_rate_pct
FROM applicants a
LEFT JOIN interns i
    ON a.applicant_id = i.applicant_id
    AND i.completion_status = 'Completed'
GROUP BY a.domain
ORDER BY conversion_rate_pct DESC;
"""

result4 = pd.read_sql(query4, conn)

result4
```

```python
# Query 5: Counts completed interns and dropouts per domain
# This query compares the number of completed interns and dropouts in each domain.
query5 = """
SELECT
    domain,
    completion_status,
    COUNT(*) AS intern_count
FROM interns
GROUP BY domain, completion_status
ORDER BY domain, completion_status;
"""

result5 = pd.read_sql(query5, conn)

result5
```
